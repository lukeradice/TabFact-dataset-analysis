```python
# imports

import pandas as pd
import numpy as np
import json
from datasets import load_dataset
```

```python
# load full cleaned dataset

with open("data/full_cleaned.json", "r", encoding="utf-8") as f:
    full_cleaned_raw_data = json.load(f)

with open("data/train_examples.json", "r", encoding="utf-8") as f:
    train_examples_raw_data = json.load(f)

def convert_to_dataframe(data, is_full_cleaned=False):
    record = []
    for table_id, details in data.items():
        statements = details[0]
        enatiled_statuses = details[1]
        caption = details[-1]
        
        if is_full_cleaned:
            pos_tags = details[2]
            for statement, entailed_status, statement_pos_tags in zip(statements, enatiled_statuses, pos_tags):
                record.append({
                    "table_id": table_id,
                    "statement": statement,
                    "entailed_status": entailed_status,
                    "caption": caption,
                    "pos_tags": statement_pos_tags
                }) 
        else:
            for statement, entailed_status in zip(statements, enatiled_statuses):
                record.append({
                    "table_id": table_id,
                    "statement": statement,
                    "entailed_status": entailed_status,
                    "caption": caption
                })
    return pd.DataFrame(record)

df_full_cleaned = convert_to_dataframe(full_cleaned_raw_data, is_full_cleaned=True)

df_train_examples = convert_to_dataframe(train_examples_raw_data, is_full_cleaned=False)

df_train_examples.head()
```
